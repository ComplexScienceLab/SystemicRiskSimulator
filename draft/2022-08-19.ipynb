```python
dict_a = {'a':10, 'b':20}
dict_b = {'c':30}
```

```python
class D(object):
    def __init__(self):
        self.d = 4
        self.__dict__.update(dict_a)
        self.__dict__.update(dict_b)
```

```python
new_D = D()
```

```python
print(new_D.a)
new_D.__dict__
```
