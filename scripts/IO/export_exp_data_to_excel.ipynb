 # 处理dataBB
 ## 使用`openpyxl`

 加载工作簿获取sheet页

```python

xlwb=xl.load_workbook(filepath_excel_exp_data)
xlws=xlwb['dataBB']
# sheet_dataBI = wb_xl['dataBI']
```

 ## 计算相关参数

```python
num_cols = xlutils.column_index_from_string('CA')
num_rows = xlws.max_row
```

 冻结首行首列、更改单元格缩进、小数位数、文字旋转

```python

# # 设置单元格缩进
# for row in xlws['A1':'CA1']:
#     for cell in row:
#         j.alignment = xl.styles.Alignment(wrapText=True) # 单元格缩进

# 补全title字段名称
xlws['F1'].value = "bankId"
xlws['A1'].value = "id"

# 设置字体
font = Font(name='Times')
for row in xlws.iter_rows(min_row = n_r_title, max_row = n_r_end, min_col = n_c_id, max_col = n_c_end):
    for cell in row:
        cell.font=font

# 设置字段名文本旋转、单元格缩进
align = Alignment(horizontal=None,vertical=None,text_rotation=90,wrap_text=True,shrink_to_fit=True,indent=0) 
for row in xlws.iter_rows(min_row = n_r_title, max_row = n_r_title, min_col = n_c_id, max_col = n_c_end):
    for cell in row:
        cell.alignment=align
        
# 设置title行高
xlws.row_dimensions[1].height=120

# 设置id列宽
for col in range(n_c_id,n_c_name,1):
    xlws.column_dimensions[xlutils.get_column_letter(col)].width=3
    
# 设置name列宽
for col in range(n_c_name,n_c_money,1):
    xlws.column_dimensions[xlutils.get_column_letter(col)].width=5
    
# 设置money列宽
for col in range(n_c_money,n_c_mark,1):
    xlws.column_dimensions[xlutils.get_column_letter(col)].width=8
    
# 设置mark列宽
for col in range(n_c_mark,n_c_list,1):
    xlws.column_dimensions[xlutils.get_column_letter(col)].width=3
    
# 设置list列宽
for col in range(n_c_list,n_c_end,1):
    xlws.column_dimensions[xlutils.get_column_letter(col)].width=12
    

# 设置money数字精度
for row in xlws.iter_rows(min_row = n_r_item, max_row = n_r_end, min_col = n_c_money, max_col = n_c_mark-1):
    for cell in row:
        cell.number_format='0.00'

# 设置mark数字精度
for row in xlws.iter_rows(min_row = n_r_item, max_row = n_r_end, min_col = n_c_mark, max_col = n_c_list-1):
    for cell in row:
        cell.number_format='0'

        
# 冻结单元格
xlws.freeze_panes = 'I2'  # 冻结首行首列

```
