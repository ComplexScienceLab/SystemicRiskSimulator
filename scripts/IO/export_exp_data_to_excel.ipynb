## 导入安装包

```python
#
# This example writes data to the existing empty dataset created by h5_crtdat.py and then reads it back.
#
import os

import h5py
import numpy as np
import xlwings as xw
import openpyxl as xl
import pandas as pd
#
# Open an existing file using default properties.
#
```

## 设置工作路径

```python
print("start")
os.chdir("/Users/ethan/LocalFiles/ResearchFile/SystemicRisk")
os.getcwd()
root_path = os.getcwd()
```

## 设置实验文件

```python
folder_exp = "test_20220527153746"
path_root_exp = os.path.join(os.getcwd(), "data/sims")
path_exp = os.path.join(path_root_exp, folder_exp)
folder_exp_output_data = "exp_output_data"
folder_exp_analyses = "analyses"
folderpath_exp_output_data = os.path.join(path_exp, folder_exp_output_data)
folderpath_exp_analyses = os.path.join(path_exp, folder_exp_analyses)
if not os.path.exists(folderpath_exp_analyses):
    os.makedirs(folderpath_exp_analyses)
filename_excel_exp_data = "exp_data.xlsx"
filepath_excel_exp_data = os.path.join(folderpath_exp_analyses, filename_excel_exp_data)
filename_exp_data_BB = "BB_exp=1.csv"
filepath_exp_data_BB = os.path.join(folderpath_exp_output_data, filename_exp_data_BB)
```

## 获取数据源

```python
df_dataBB = pd.read_csv(filepath_exp_data_BB)
```

# 使用`xlwings`

## 创建实验后的Excel表格文件

```python
book = xw.Book()
book.save(filepath_excel_exp_data)
book.close()
```

## 打开已经建好的Excel表格文件

```python
app = xw.App(visible=True, add_book=False)
app.display_alerts = False  # 不显示Excel消息框
book = app.books.open(filepath_excel_exp_data)
```

## 创建表
创建表dataBB、dataBI，用于显示相关数据。

```python
sheet_dataBB = book.sheets[0]
sheet_dataBB.name = "dataBB"
sheet_dataBI = book.sheets.add("dataBI")
book.save()
```

## 加载相关数据

表格载入数据

```python

sheet_dataBB.range('A1').value = df_dataBB
```

设置单元格样式

```python
sheet_dataBB.range('A1:CA41').color=None
sheet_dataBB.range('A1:CA41').font.name="Times"
sheet_dataBB.range('A1').value="id"
book.save()
```

# 使用`openpyxl`

## 加载表格

```python
book=xl.load_workbook(filepath_excel_exp_data)
sheet=book['dataBB']
print(sheet['A1'].value)
```

## 退出处理Excel表格

```python
book.save()
book.close()
app.quit()
```
